```python
# Step 8 - Process Every Line

messages = []

for line in chat:

    if ": " not in line:
        continue

    try:

        part1, part2 = line.split(" - ")

        date, time = part1.split(", ")

        sender, message = part2.split(": ", 1)

        messages.append({
            "date": date,
            "time": time,
            "sender": sender,
            "message": message.strip()
        })

    except:
        pass
```

## Feature 4 - Activity Heatmap (NumPy)


```python
# Step 1 — Import NumPy
import numpy as np
```

```python
# Step 2 — Create the heatmap

heatmap = np.zeros((len(participants), 24), dtype=int)

print(heatmap.shape)
```

```python
# Step 3 — Create participant row numbers

person_index = {}

for i, person in enumerate(participants):
    person_index[person] = i

print(person_index)
```

```python
# Step 4 — Fill the heatmap

for msg in messages:
    row = person_index[msg["sender"]]
    hour = int(msg["time"].split(":")[0])

    heatmap[row][hour] += 1
```

```python
# Step 5 — Check the result
print(heatmap)
```

```python
# Step 6 — Print the required heatmap

print("ACTIVITY HEATMAP (messages by hour)")
print()

# Print hour headings
print("        ", end="")

for hour in range(0, 24, 3):
    print(f"{hour:02}   ", end="")

print()

# Print each participant
for i, person in enumerate(participants):

    # Find this person's maximum messages in any hour
    person_max = np.max(heatmap[i])

    print(f"{person:<8}", end="")

    for hour in range(0, 24, 3):

        value = heatmap[i][hour]

        # Avoid division by zero
        if person_max == 0:
            symbol = "."
        else:
            percentage = value / person_max

            if percentage <= 0.25:
                symbol = "."
            elif percentage <= 0.50:
                symbol = "░"
            elif percentage <= 0.75:
                symbol = "▓"
            else:
                symbol = "█"

        print(f"{symbol}    ", end="")

    print()
```

# Feature 6: Response Speed & Silent Streaks.

# This feature has 2 parts:
        # 1> Average response time for each person.
        # 2> Longest silent streak for each person.

```python
# Step 1 — Convert each message into a datetime

from datetime import datetime

for msg in messages:
    msg["datetime"] = datetime.strptime(
        msg["date"] + " " + msg["time"],
        "%d/%m/%y %H:%M"
    )

print(messages[0])
```

```python
# Step 2 — Find each person's response time

response_times = {}

for i in range(1, len(messages)):
    previous = messages[i - 1]
    current = messages[i]

    if previous["sender"] != current["sender"]:
        person = current["sender"]

        gap = current["datetime"] - previous["datetime"]

        seconds = gap.total_seconds()

        if person not in response_times:
            response_times[person] = []

        response_times[person].append(seconds)

print(response_times)
```

```python
# Step 3 —Calculate the average response time for each person.

average_response = {}

for person in response_times:
    total = sum(response_times[person])
    count = len(response_times[person])

    average_response[person] = total / count

print(average_response)
```

```python
# Step 4 — Find the fastest and slowest replier.


# Now we compare everyone's average response time.

fastest_person = min(average_response, key=average_response.get)
slowest_person = max(average_response, key=average_response.get)

fastest_time = average_response[fastest_person]
slowest_time = average_response[slowest_person]

print("Fastest replier :", fastest_person)
print("Average time :", fastest_time, "seconds")

print("Slowest replier :", slowest_person)
print("Average time :", slowest_time, "seconds")
```

```python
# Step 5: Count messages for each person on each day



daily_messages = {}

for msg in messages:
    person = msg["sender"]
    date = msg["date"]

    if person not in daily_messages:
        daily_messages[person] = {}

    if date not in daily_messages[person]:
        daily_messages[person][date] = 0

    daily_messages[person][date] += 1

print(daily_messages)
```

```python
# Step 6 — find the silent days.

from datetime import datetime, timedelta
silent_days = {}

current = start

while current <= end:

    date = current.strftime("%d/%m/%y")

    for person in participants:

        if person not in daily_messages:
            daily_messages[person] = {}

        if date not in daily_messages[person]:
            daily_messages[person][date] = 0

    current = current + timedelta(days=1)

print(daily_messages)
```

```python
# Step 7 - Find longest silent streak

silent_streaks = {}

for person in participants:

    current_streak = 0
    longest_streak = 0

    current = start

    while current <= end:

        date = current.strftime("%d/%m/%y")

        if daily_messages[person][date] == 0:
            current_streak += 1

            if current_streak > longest_streak:
                longest_streak = current_streak
        else:
            current_streak = 0

        current = current + timedelta(days=1)

    silent_streaks[person] = longest_streak

print(silent_streaks)
```

```python
# Step 8 - Final Feature 6 Output

fastest_minutes = fastest_time / 60
slowest_minutes = slowest_time / 60

print("RESPONSE PATTERNS")
print()
print(f"Fastest replier : {fastest_person} (avg {fastest_minutes:.1f} minutes)")
print(f"Slowest replier : {slowest_person} (avg {slowest_minutes:.1f} minutes)")

print()
print("LONGEST SILENT STREAKS")

for person, days in sorted(silent_streaks.items(), key=lambda x: x[1], reverse=True):
    print(f"{person:<8}: {days} days")
```

## Feature 7 — Personality Archetype Detection.


```python
#step 1: average_burst → Spammer
#step 2 : caring_count → Group Mom
# step 3 :night_percentage → Night Owl
# step 4 : average_words → Storyteller
# step 5 : caps_percentage → Drama Queen
# step 6 :silent_day_percentage → Ghost
# step 7 : emoji_percentage → Emoji Lover
```

```python
# Step 1: THE SPAMMER
# Calculate average consecutive message burst for each person

burst_data = {}

current_person = None
current_burst = 0

for msg in messages:

    person = msg["sender"]

    if person == current_person:
        current_burst += 1

    else:
        if current_person is not None:

            if current_person not in burst_data:
                burst_data[current_person] = []

            burst_data[current_person].append(current_burst)

        current_person = person
        current_burst = 1


# Add the final burst
if current_person is not None:

    if current_person not in burst_data:
        burst_data[current_person] = []

    burst_data[current_person].append(current_burst)


# Calculate average burst
average_burst = {}

for person in burst_data:
    average_burst[person] = sum(burst_data[person]) / len(burst_data[person])

print(average_burst)
```

```python
# Step 2 — Count caring keywords

# Feature 7 - Step 2
# Count caring keywords for each person

caring_keywords = [
    "okay",
    "safe",
    "eat",
    "sleep",
    "take care",
    "are you",
    "please",
    "reminder",
    "drink water",
    "don't forget"
]

caring_count = {}

for msg in messages:

    person = msg["sender"]
    text = msg["message"].lower()

    if person not in caring_count:
        caring_count[person] = 0

    for keyword in caring_keywords:
        if keyword in text:
            caring_count[person] += 1

print(caring_count)
```

```python
# Step 3 — THE NIGHT OWL

# Feature 7 - Step 3
# Calculate night-time message percentage

night_messages = {}
total_messages = {}

for msg in messages:

    person = msg["sender"]
    hour = int(msg["time"].split(":")[0])

    # Count total messages
    if person not in total_messages:
        total_messages[person] = 0

    total_messages[person] += 1

    # Check if message was sent at night
    if hour >= 23 or hour <= 4:

        if person not in night_messages:
            night_messages[person] = 0

        night_messages[person] += 1


# Calculate percentage
night_percentage = {}

for person in participants:

    night_count = night_messages.get(person, 0)
    total_count = total_messages[person]

    night_percentage[person] = (night_count / total_count) * 100

print(night_percentage)
```

```python
# Step 4 — THE STORYTELLE

# Feature 7 - Step 4
# Calculate average words per message

word_totals = {}
message_totals = {}

for msg in messages:

    person = msg["sender"]
    text = msg["message"]

    words = text.split()
    word_count = len(words)

    if person not in word_totals:
        word_totals[person] = 0

    if person not in message_totals:
        message_totals[person] = 0

    word_totals[person] += word_count
    message_totals[person] += 1


# Calculate average
average_words = {}

for person in participants:

    average_words[person] = (
        word_totals[person] / message_totals[person]
    )

print(average_words)
```

```python
# Step 5 — THE DRAMA QUEEN

# Feature 7 - Step 5
# Calculate ALL-CAPS message percentage

caps_messages = {}
valid_messages = {}

for msg in messages:

    person = msg["sender"]
    text = msg["message"].strip()

    if person not in caps_messages:
        caps_messages[person] = 0

    if person not in valid_messages:
        valid_messages[person] = 0

    # Ignore messages shorter than 3 characters
    if len(text) < 3:
        continue

    valid_messages[person] += 1

    # Check if the message is completely uppercase
    if text.isupper():
        caps_messages[person] += 1


# Calculate percentage
caps_percentage = {}

for person in participants:

    if valid_messages[person] > 0:
        caps_percentage[person] = (
            caps_messages[person] / valid_messages[person]
        ) * 100
    else:
        caps_percentage[person] = 0

print(caps_percentage)
```

```python
# Now Step 6 — THE GHOST

# Feature 7 - Step 6
# Calculate silent-day percentage

silent_day_percentage = {}

total_days = (end - start).days + 1

for person in participants:

    silent_days_count = 0

    current = start

    while current <= end:

        date = current.strftime("%d/%m/%y")

        if daily_messages[person][date] == 0:
            silent_days_count += 1

        current = current + timedelta(days=1)

    silent_day_percentage[person] = (
        silent_days_count / total_days
    ) * 100

print(silent_day_percentage)
```

```python
# Step 7 — THE EMOJI LOVER 😄

# Feature 7 - Step 7
# Count emoji usage

emoji_count = {}
total_messages_person = {}

for msg in messages:

    person = msg["sender"]
    text = msg["message"]

    if person not in emoji_count:
        emoji_count[person] = 0

    if person not in total_messages_person:
        total_messages_person[person] = 0

    total_messages_person[person] += 1

    for char in text:

        code = ord(char)

        if (
            0x1F300 <= code <= 0x1FAFF
            or 0x2600 <= code <= 0x27BF
        ):
            emoji_count[person] += 1

print(emoji_count)
```

```python
# Step 8 — THE EMOJI LOVER: Calculate emoji usage percentage


emoji_percentage = {}

for person in participants:

    emoji_percentage[person] = (
        emoji_count[person] / total_messages_person[person]
    ) * 100

print(emoji_percentage)
```

```python
print("AVERAGE BURST:")
print(average_burst)

print("\nCARING KEYWORDS:")
print(caring_count)

print("\nNIGHT MESSAGE %:")
print(night_percentage)

print("\nAVERAGE WORDS:")
print(average_words)

print("\nALL-CAPS %:")
print(caps_percentage)

print("\nSILENT-DAY %:")
print(silent_day_percentage)

print("\nEMOJI USAGE:")
print(emoji_percentage)
```

```python
# step 7

funny_words = ["lol", "lmao", "haha", "rofl", "lmfao"]

funny_count = {}

for person in participants:
    funny_count[person] = 0

for msg in messages:
    person = msg["sender"]
    text = msg["message"].lower()
    words = text.split()

    for word in words:
        word = word.strip(".,!?;:'\"()[]{}-_")

        if word in funny_words:
            funny_count[person] += 1

comedian_percentage = {}

for person in participants:
    comedian_percentage[person] = (
        funny_count[person] / total_messages_person[person]
    ) * 100

print("FUNNY WORD COUNT:")
print(funny_count)

print("\nCOMEDIAN %:")
print(comedian_percentage)
```

```python
# Step 8: THE QUESTION MASTER ❓

# Feature 7 - Step 8
# THE QUESTION MASTER

question_count = {}

for person in participants:
    question_count[person] = 0

for msg in messages:
    person = msg["sender"]
    text = msg["message"].strip()

    if text.endswith("?"):
        question_count[person] += 1

question_percentage = {}

for person in participants:
    question_percentage[person] = (
        question_count[person] / total_messages_person[person]
    ) * 100

print("QUESTION COUNT:")
print(question_count)

print("\nQUESTION %:")
print(question_percentage)
```

```python
# Feature 7 - Final Personality Assignment

personality_scores = {}
personality_result = {}

# Maximum values for normalization
max_burst = max(average_burst.values())
max_caring = max(caring_count.values())
max_night = max(night_percentage.values())
max_words = max(average_words.values())
max_caps = max(caps_percentage.values())
max_silent = max(silent_day_percentage.values())
max_comedian = max(comedian_percentage.values())
max_questions = max(question_percentage.values())


for person in participants:

    scores = {}

    # 1. THE SPAMMER
    if average_burst[person] > 3:
        scores["THE SPAMMER"] = (
            average_burst[person] / max_burst
        ) * 100
    else:
        scores["THE SPAMMER"] = 0

    # 2. THE GROUP MOM
    scores["THE GROUP MOM"] = (
        caring_count[person] / max_caring
    ) * 100

    # 3. THE NIGHT OWL
    if night_percentage[person] > 60:
        scores["THE NIGHT OWL"] = (
            night_percentage[person] / max_night
        ) * 100
    else:
        scores["THE NIGHT OWL"] = 0

    # 4. THE STORYTELLER
    if average_words[person] > 30:
        scores["THE STORYTELLER"] = (
            average_words[person] / max_words
        ) * 100
    else:
        scores["THE STORYTELLER"] = 0

    # 5. THE DRAMA QUEEN
    if caps_percentage[person] > 30:
        scores["THE DRAMA QUEEN"] = (
            caps_percentage[person] / max_caps
        ) * 100
    else:
        scores["THE DRAMA QUEEN"] = 0

    # 6. THE GHOST
    if silent_day_percentage[person] > 60:
        scores["THE GHOST"] = (
            silent_day_percentage[person] / max_silent
        ) * 100
    else:
        scores["THE GHOST"] = 0

    # 7. THE COMEDIAN
    if max_comedian > 0:
        scores["THE COMEDIAN"] = (
            comedian_percentage[person] / max_comedian
        ) * 100
    else:
        scores["THE COMEDIAN"] = 0

    # 8. THE QUESTION MASTER
    if question_percentage[person] > 25:
        scores["THE QUESTION MASTER"] = (
            question_percentage[person] / max_questions
        ) * 100
    else:
        scores["THE QUESTION MASTER"] = 0

    # Save all scores
    personality_scores[person] = scores

    # Highest score wins
    # If there is a tie, the first archetype wins.
    personality_result[person] = max(
        scores,
        key=scores.get
    )


# Print final personalities
print("PERSONALITY ARCHETYPES")
print("-" * 50)

for person in participants:
    print(f"{person:<10} → {personality_result[person]}")
```
